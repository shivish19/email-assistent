```python
import os
from dotenv import load_dotenv

# ✅ GmailToolkit moved from langchain_community → langchain_google_community
from langchain_google_community import GmailToolkit
from langchain_google_community.gmail.utils import (
    build_resource_service,
    get_gmail_credentials,
)

# import langchainhub as hub
from langchain_google_genai import ChatGoogleGenerativeAI 


# ✅ create_openai_functions_agent is deprecated — use langgraph's create_react_agent
from langgraph.prebuilt import create_react_agent


```

```python
# Load environment variables from .env file
load_dotenv()

# Ensure the OPENAI_API_KEY environment variable is set
api_key = os.getenv("GOOGLE_API_KEY")
if not api_key:
    raise ValueError("The OPENAI_API_KEY environment variable is not set")
```

```python
# Set up Gmail credentials
credentials = get_gmail_credentials(
    token_file="token.json",
    scopes=["https://mail.google.com/"],
    client_sercret_file="credentials.json",
)

# Build API resource service for Gmail
api_resource = build_resource_service(credentials=credentials)

# Initialize Gmail toolkit with the API resource
toolkit = GmailToolkit(api_resource=api_resource)
tools = toolkit.get_tools()
```

```python

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",  # free tier
    google_api_key=os.getenv("GOOGLE_API_KEY"),
    temperature=0
)
agent_executor = create_react_agent(
    llm,
    toolkit.get_tools(),
    prompt="You are a helpful Gmail assistant. You can read, search, draft and send emails on behalf of the user. Always confirm before sending any email."
)

# # Initialize the AgentExecutor with the agent and tools
# agent_executor = AgentExecutor(
#     agent=agent,
#     tools=toolkit.get_tools(),
#     verbose=False,  # Set to False to prevent sensitive email information from being displayed
# )

# agent_executor = create_react_agent(
#     llm,
#     tools,
#     prompt="You are a helpful Gmail assistant."
# )
```

```python
result = agent_executor.invoke({
    "messages": [("user", "which model do u use")]
})
print(result["messages"][-1].content)
```
